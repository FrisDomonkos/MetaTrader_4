```python
import pandas as pd

from PATHS import DATA_FOLDER
from data_prep import merge_open_close
from info_extraction import calc_profit
```

```python
# J = jump trailing
# S = simple trailing
# N = adt-atr filter NOT used
# L,G = adt-atr filter used and atr <, > parameter amount


open_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/simple_trailing/open_trades_test.csv',sep=';')
close_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/simple_trailing/closed_trades_test.csv',sep=';')
S_N_merge_df = merge_open_close(open_df,close_df)

open_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/simple_trailing/ADX_ATR_FILTER/ATR_FILTER_GREATERTHAN_VALUE/open_trades_test.csv',sep=';')
close_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/simple_trailing/ADX_ATR_FILTER/ATR_FILTER_GREATERTHAN_VALUE/closed_trades_test.csv',sep=';')
S_G_merge_df = merge_open_close(open_df,close_df)

open_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/simple_trailing/ADX_ATR_FILTER/ATR_FILTER_LESSTHAN_VALUE/open_trades_test.csv',sep=';')
close_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/simple_trailing/ADX_ATR_FILTER/ATR_FILTER_LESSTHAN_VALUE/closed_trades_test.csv',sep=';')
S_L_merge_df = merge_open_close(open_df,close_df)

open_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/jump_trailing/open_trades_test.csv',sep=';')
close_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/jump_trailing/closed_trades_test.csv',sep=';')
J_N_merge_df = merge_open_close(open_df,close_df)

open_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/jump_trailing/ADX_ATR_FILTER/ATR_FILTER_GREATERTHAN_VALUE/open_trades_test.csv',sep=';')
close_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/jump_trailing/ADX_ATR_FILTER/ATR_FILTER_GREATERTHAN_VALUE/closed_trades_test.csv',sep=';')
J_G_merge_df = merge_open_close(open_df,close_df)

open_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/jump_trailing/ADX_ATR_FILTER/ATR_FILTER_LESSTHAN_VALUE/open_trades_test.csv',sep=';')
close_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/jump_trailing/ADX_ATR_FILTER/ATR_FILTER_LESSTHAN_VALUE/closed_trades_test.csv',sep=';')
J_L_merge_df = merge_open_close(open_df,close_df)

df_dict = {
    'SimpleTrailing_NoFilter_None':[S_N_merge_df],
    'SimpleTrailing_Filter_Greater':[S_G_merge_df],
    'SimpleTrailing_Filter_Less':[S_L_merge_df],
    'JumpTrailing_NoFilter_None':[J_N_merge_df],
    'JumpTrailing_Filter_Greater':[J_G_merge_df],
    'JumpTrailing_Filter_Less':[J_L_merge_df]
    }

for key in df_dict:
    profit_df = calc_profit(df_dict[key][0],len(df_dict[key][0]))
    profit_df['Profit_With_Drawdown'] = profit_df['Profit'] * (30 / profit_df['Max_Drawdown_Pct'])
    df_dict[key].append(profit_df)

'''for key in df_dict:
    temp_df = df_dict[key][0]
    Mask = (
        (temp_df["slow_MA"] > 0.0)
        &
        (
            (temp_df["SAR"] <= 0.0)
            |
            (temp_df["fast_MA_prev"] > 0.0)
        )
    )
    df_dict[key].append(calc_profit(temp_df,
                                    len(temp_df),
                                    MASK))
    '''
```
