Evaluation of test runs by hand.

```python
import pandas as pd
import numpy as np
from PATHS import DATA_FOLDER
from data_prep import merge_open_close, assign_candle_id
from info_extraction import calc_profit
```

```python
open_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/jump_trailing/open_trades_test.csv',sep=';')
close_df = pd.read_csv(DATA_FOLDER + 'M15/2024-2026/smaller_spike/jump_trailing/closed_trades_test.csv',sep=';')
base_merge_df = merge_open_close(open_df,close_df)

df_dict = {
    'Base':[base_merge_df],
    }

for key in df_dict:
    profit_df = calc_profit(df_dict[key][0],len(df_dict[key][0]))
    profit_df['Profit_With_Drawdown'] = profit_df['Profit'] * (30 / profit_df['Max_Drawdown_Pct'])
    df_dict[key].append(profit_df)
```

```python
# Order trades by time (without grouping by test_id)
df = df_dict['Base'][0]
df["OpenTime"] = pd.to_datetime(df["OpenTime"], format="%Y.%m.%d %H:%M:%S")

# sort by datetime
df = df.sort_values(by="OpenTime")

# (optional) reset index
df = df.reset_index(drop=True)
```
